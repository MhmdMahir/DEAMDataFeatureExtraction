```python
import pandas as pd
import numpy as np

print("Libraries imported")
```

```python
file1 = "/content/sample_data/metadata1.csv"
file2 = "/content/sample_data/metadata2.csv"
file3 = "/content/sample_data/metadata3.csv"



print("files are uploaded")
```

```python
df_1 = pd.read_csv(file1)
df_2 = pd.read_csv(file2)
df_3 = pd.read_csv(file3)


print(df_1.shape)
print(df_2.shape)
print(df_3.shape)
```

```python
df = [df_1,df_2,df_3]


for df in df:
  print(df.columns)
```

```python
shared_column = ["id", "artist", "title", "genre"]
```

```python
df_1 = df_1.rename(columns=
 {
    "song_id":shared_column[0],
    "Artist": shared_column[1],
    "Song title": shared_column[2],
    "Genre": shared_column[3]
  }
)

df_1 = df_1[shared_column]


unwanted_symbols = "\t"
df_1 = df_1.replace(r'\t', '', regex=True)
df_1
```

```python
df_2 = df_2.rename(columns=
 {
    "Id":shared_column[0],
    "Artist": shared_column[1],
    "Track": shared_column[2],
    "Genre": shared_column[3]
  }
)

df_2 = df_2[shared_column]
df_2
```

```python
df_3 = df_3.rename(columns=
 {
    "id":shared_column[0],
    "artist": shared_column[2],
    "title": shared_column[1],
    "genre": shared_column[3]
  }
)

df_3 = df_3[shared_column]
df_3
```

```python
df = pd.concat([df_1, df_2,df_3], ignore_index=True)
df.tail(500)


df.to_json("combinedmetadata.csv")
```
