```python
arousal = pd.read_csv("/content/sample_data/arousal.csv")
valence = pd.read_csv("/content/sample_data/valence.csv")

```

```python
missing_song_ids = missing_rows['id']

arousal_fil = arousal[~arousal['song_id'].isin(missing_song_ids)]
valence_fil = valence[~valence['song_id'].isin(missing_song_ids)]
```

```python
arousal_fil.head()
```

```python
print('Shape of arousal_fil:', arousal_fil.shape)
print('Shape of valence_fil:', valence_fil.shape)
```

```python
arousal_mean = arousal_fil.groupby('song_id').mean().mean(axis=1).rename('arousal_mean')
valence_mean = valence_fil.groupby('song_id').mean().mean(axis=1).rename('valence_mean')
```

```python
# Drop any existing redundant mean columns before merging to ensure a clean state
columns_to_drop = [col for col in merged_df.columns if 'valence_mean' in col or 'arousal_mean' in col]
merged_df = merged_df.drop(columns=columns_to_drop)

# Merge the correctly named valence_mean Series
merged_df = merged_df.merge(
    valence_mean,
    left_on='id',
    right_index=True,
    how='left'
)

# Merge the correctly named arousal_mean Series
merged_df = merged_df.merge(
    arousal_mean,
    left_on='id',
    right_index=True,
    how='left'
)

display(merged_df.head())
```

```python
merged_df.to_json("metadata.json", orient='records', indent=4)
```

```python
# Filtering for rows where -1 > Arousal > 1 and -1 > Valence > 1 (mathematically impossible)
extremely_out_of_range_data = merged_df[
    ((-1 > merged_df['arousal_mean']) & (merged_df['arousal_mean'] > 1)) |
    ((-1 > merged_df['valence_mean']) & (merged_df['valence_mean'] > 1))
]
display(extremely_out_of_range_data)
```

As you can see, the above DataFrame is empty. This is because a number cannot be simultaneously less than -1 and greater than 1.

If you intended to find rows where Arousal or Valence are *outside* the range of -1 to 1 (i.e., less than -1 OR greater than 1), here is the corrected code:

```python
# Filtering for rows where Arousal < -1 or Arousal > 1
# AND Valence < -1 or Valence > 1

out_of_range_data = merged_df[
    ((merged_df['arousal_mean'] < -1) | (merged_df['arousal_mean'] > 1)) |
    ((merged_df['valence_mean'] < -1) | (merged_df['valence_mean'] > 1))
]

display(out_of_range_data)
```
